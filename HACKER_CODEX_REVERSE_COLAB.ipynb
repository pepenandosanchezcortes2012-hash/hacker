```python
# ==============================================================================
# MÓDULO 05: INYECCIÓN DE ARQUITECTURA DE SOFTWARE
# ==============================================================================
import os
os.makedirs("/content/colab_codex_core", exist_ok=True)

# 1. KEEPALIVE DAEMON
with open("/content/colab_codex_core/keepalive_daemon.py", "w") as f:
    f.write('''import os, sys, time, subprocess, logging
from datetime import datetime
from pathlib import Path

LOG_PATH = Path("/content/keepalive_daemon.log")
DRIVE_PULSE_PATH = Path("/content/drive/MyDrive/HACKER_CODEX_CORE/logs/heartbeat.txt")

logging.basicConfig(level=logging.INFO, format="%(asctime)s [KEEPALIVE] %(message)s", handlers=[logging.FileHandler(LOG_PATH)])
logger = logging.getLogger("KeepAlive")

def gpu_micro_pulse():
    try:
        import torch
        if torch.cuda.is_available():
            dev = torch.device("cuda:0")
            a = torch.ones((128, 128), device=dev)
            b = torch.ones((128, 128), device=dev)
            _ = torch.matmul(a, b)
            torch.cuda.synchronize(dev)
            return True
    except Exception as e:
        logger.error(f"Error GPU pulse: {e}")
    return False

def disk_heartbeat():
    try:
        if DRIVE_PULSE_PATH.parent.exists():
            with open(DRIVE_PULSE_PATH, "a") as df:
                df.write(f"VITAL_PULSE_OK: {datetime.now().isoformat()}\n")
    except Exception: pass

while True:
    try:
        gpu_micro_pulse()
        disk_heartbeat()
        time.sleep(45)
    except Exception:
        time.sleep(10)
''')

# 2. REVERSING ARSENAL
with open("/content/colab_codex_core/reversing_arsenal.py", "w") as f:
    f.write('''import os, sys, math, subprocess
from typing import Dict, Any, List

class ReversingArsenal:
    @staticmethod
    def inspect_file(filepath: str) -> Dict[str, Any]:
        if not os.path.exists(filepath): return {"error": "Archivo no existe"}
        p = subprocess.run(["file", "-b", filepath], capture_output=True, text=True)
        return {"path": filepath, "type": p.stdout.strip(), "size": os.path.getsize(filepath)}

    @staticmethod
    def run_checksec(filepath: str) -> Dict[str, str]:
        try:
            from pwn import ELF
            elf = ELF(filepath, checksec=False)
            return {
                "RELRO": str(elf.relro),
                "Canary": "Enabled" if elf.canary else "Disabled",
                "NX": "Enabled" if elf.nx else "Disabled",
                "PIE": "Enabled" if elf.pie else "Disabled",
                "Arch": f"{elf.arch} ({elf.bits} bits)"
            }
        except Exception as e:
            return {"error": str(e)}

    @staticmethod
    def calculate_entropy(filepath: str) -> float:
        if not os.path.exists(filepath): return 0.0
        with open(filepath, "rb") as f: data = f.read()
        if not data: return 0.0
        counts = [0]*256
        for b in data: counts[b] += 1
        ent = 0.0
        tot = len(data)
        for c in counts:
            if c > 0:
                p = c / tot
                ent -= p * math.log2(p)
        return round(ent, 4)

    @staticmethod
    def disassemble_raw_bytes(hex_bytes: str, arch: str = "x86_64") -> List[str]:
        try:
            from capstone import Cs, CS_ARCH_X86, CS_MODE_64, CS_MODE_32, CS_ARCH_ARM64, CS_MODE_ARM
            clean_hex = hex_bytes.replace(" ", "").replace("\\x", "").strip()
            code = bytes.fromhex(clean_hex)
            md = Cs(CS_ARCH_X86, CS_MODE_64) if arch == "x86_64" else Cs(CS_ARCH_ARM64, CS_MODE_ARM)
            return [f"0x{i.address:x}:\t{i.mnemonic}\t{i.op_str}" for i in md.disasm(code, 0x1000)]
        except Exception as e:
            return [f"Error Capstone: {e}"]

    @staticmethod
    def assemble_asm(asm_code: str, arch: str = "x86_64") -> Dict[str, Any]:
        try:
            from keystone import Ks, KS_ARCH_X86, KS_MODE_64
            ks = Ks(KS_ARCH_X86, KS_MODE_64)
            encoding, count = ks.asm(asm_code)
            raw = bytes(encoding)
            return {"hex": raw.hex(), "c_string": "".join([f"\\x{b:02x}" for b in raw]), "count": count}
        except Exception as e:
            return {"error": str(e)}
''')

print("✅ Módulos de soporte inyectados en /content/colab_codex_core/.")

```
